```python
# 59. Find rows where marks is missing.
sm_df[sm_df["marks"].isnull()]
```

```python
# 60. Find the percentage of missing values in each column.
sm_df.isnull().mean() * 100
```

```python
# 61. Fill missing marks with the mean marks.
sm_df["marks"].fillna(sm_df["marks"].mean())
```

```python
# 62. Fill missing phone values with 'Not Available'.
sm_df["phone"].fillna("Not Available")
```

```python
# 63. Drop rows where marks is missing.
sm_df = sm_df.dropna(subset=["marks"])
sm_df
```

```python
# 64. Drop a column only if you decide it is not useful for analysis.
sm_df = sm_df.drop(columns=["column_name"])
```

```python
# 65. Replace missing values in more than one column using a dictionary.
sm_df = sm_df.fillna({
    "marks": sm_df["marks"].mean(),
    "age": sm_df["age"].mean()
})
sm_df
```

```python
# 66. Create a cleaned copy of the DataFrame without modifying the original.
cleaned_df = sm_df.copy()
cleaned_df
```

```python
# 7. Check again that the cleaned DataFrame contains no missing values.
cleaned_df.isnull().sum()
```

11. Merge, Join & Concat

```python
# 125. Merge employees.csv with departments.csv using department.
d_data = pd.read_csv("departments.csv")
d_df = pd.DataFrame(d_data)

pd.merge(e_df,d_df,on="department")
```

```python
# 126. Perform an inner merge and inspect the result.
pd.merge(e_df, d_df, on="department", how="inner")
```

```python
# 127. Perform a left merge and identify whether any employees have no matching department.
pd.merge(e_df,d_df,on="department",how="left")
```

```python
# 128. Perform a right merge.
pd.merge(e_df,d_df,on= "department",how="right")
```

```python
# 129. Perform an outer merge.
pd.merge(e_df,d_df,on= "department",how="outer")
```

```python
# 130. Use suffixes when two DataFrames contain the same non-key column name.
pd.merge(e_df,d_df,on="department" ,suffixes=("_employee", "_department"))
```

```python
# 131. Set department as the index in both DataFrames and use join().
e_df = e_df.set_index("department")
d_df = d_df.set_index("department")

result = e_df.join(d_df)
result
```
